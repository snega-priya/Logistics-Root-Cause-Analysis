```python

import pandas as pd
import numpy as np

# Make results reproducible
np.random.seed(42)

# Number of shipments
n = 30000

# Create shipment records
shipments = pd.DataFrame({
    "Shipment_ID": [f"S{i:05d}" for i in range(1, n + 1)],

    "Hub_ID": np.random.choice(
        [f"H{i:02d}" for i in range(1, 13)], n
    ),

    "Vehicle_ID": np.random.choice(
        [f"V{i:03d}" for i in range(1, 51)], n
    ),

    "Route_ID": np.random.choice(
        [f"R{i:02d}" for i in range(1, 31)], n
    ),

    "Customer_ID": np.random.choice(
        [f"C{i:03d}" for i in range(1, 501)], n
    ),

    "Service_Type": np.random.choice(
        ["Standard", "Express", "Economy"], n
    ),

    "Booking_Date": pd.to_datetime(
        np.random.choice(
            pd.date_range("2026-01-01", "2026-06-30"),
            n
        )
    ),

    "Delivery_Time_Hours": np.random.randint(2, 25, n),

    "Promised_Time_Hours": np.random.randint(5, 20, n)
})

# Calculate actual delivery time
shipments["Actual_Delivery_Hours"] = (
    shipments["Delivery_Time_Hours"]
)

# Determine whether delivery was on time
shipments["On_Time"] = (
    shipments["Actual_Delivery_Hours"]
    <= shipments["Promised_Time_Hours"]
)

print(shipments.head())
print(shipments.shape)
```

```python

vehicles = pd.DataFrame({
    "Vehicle_ID": [f"V{i:03d}" for i in range(1, 51)],

    "Vehicle_Age": np.random.randint(1, 16, 50),

    "Vehicle_Type": np.random.choice(
        ["Mini Truck", "Medium Truck", "Heavy Truck"], 50
    ),

    "Capacity_Tons": np.random.choice(
        [2, 5, 10, 15], 50
    )
})

print(vehicles.head())
print(vehicles.shape)
```

```python

scan_records = []

for i in range(len(shipments)):
    shipment_id = shipments.loc[i, "Shipment_ID"]
    booking = shipments.loc[i, "Booking_Date"]
    hub = shipments.loc[i, "Hub_ID"]

    scan_records.append({
        "Shipment_ID": shipment_id,
        "Hub_ID": hub,
        "Event_Type": "ARRIVED",
        "Scan_Time": booking + pd.Timedelta(hours=1)
    })

    scan_records.append({
        "Shipment_ID": shipment_id,
        "Hub_ID": hub,
        "Event_Type": "PROCESSED",
        "Scan_Time": booking + pd.Timedelta(hours=3)
    })

    scan_records.append({
        "Shipment_ID": shipment_id,
        "Hub_ID": hub,
        "Event_Type": "DISPATCHED",
        "Scan_Time": booking + pd.Timedelta(hours=4)
    })

warehouse_scans = pd.DataFrame(scan_records)

print(warehouse_scans.head())
print(warehouse_scans.shape)
```

```python

import pandas as pd
import numpy as np

np.random.seed(42)

# 1. Create shipment records
n = 30000

shipments = pd.DataFrame({
    "Shipment_ID": [f"S{i:05d}" for i in range(n)],
    "Hub_ID": np.random.choice([f"H{i:02d}" for i in range(1, 13)], n),
    "Vehicle_ID": np.random.choice([f"V{i:03d}" for i in range(1, 51)], n),
    "Route_ID": np.random.choice([f"R{i:02d}" for i in range(1, 31)], n),
    "Customer_ID": np.random.choice([f"C{i:03d}" for i in range(1, 501)], n),
    "Service_Type": np.random.choice(
        ["Standard", "Express", "Economy"], n,
        p=[0.55, 0.25, 0.20]
    ),
    "Booking_Date": pd.to_datetime(
        np.random.choice(pd.date_range("2026-01-01", "2026-06-30"), n)
    )
})

# 2. Create vehicle records
vehicles = pd.DataFrame({
    "Vehicle_ID": [f"V{i:03d}" for i in range(1, 51)],
    "Vehicle_Age": np.random.randint(1, 16, 50),
    "Vehicle_Type": np.random.choice(
        ["Mini Truck", "Medium Truck", "Heavy Truck"], 50
    ),
    "Capacity_Tons": np.random.choice([2, 5, 10, 15], 50)
})

# 3. Create route records
routes = pd.DataFrame({
    "Route_ID": [f"R{i:02d}" for i in range(1, 31)],
    "Distance_KM": np.random.randint(10, 250, 30),
    "Traffic_Level": np.random.choice(["Low", "Medium", "High"], 30),
    "Route_Type": np.random.choice(["Urban", "Highway", "Mixed"], 30)
})

# 4. Create hub records
hub_names = [
    "Kolkata Central", "Howrah", "Salt Lake", "New Town",
    "Dum Dum", "Bally", "Barasat", "Behala",
    "Tollygunge", "Rajarhat", "Park Street", "Barrackpore"
]

hubs = pd.DataFrame({
    "Hub_ID": [f"H{i:02d}" for i in range(1, 13)],
    "Hub_Name": hub_names,
    "Hub_Capacity": np.random.randint(500, 2000, 12)
})

# 5. Integrate all records
analysis_data = shipments.merge(vehicles, on="Vehicle_ID")
analysis_data = analysis_data.merge(routes, on="Route_ID")
analysis_data = analysis_data.merge(hubs, on="Hub_ID")

# 6. Add operational factors
analysis_data["Post_Period"] = (
    analysis_data["Booking_Date"] >= "2026-04-01"
).astype(int)

analysis_data["High_Traffic"] = (
    analysis_data["Traffic_Level"] == "High"
).astype(int)

analysis_data["High_Volume"] = (
    analysis_data["Booking_Date"].dt.isocalendar().week % 4 == 0
).astype(int)

# 7. Simulate warehouse processing time
hub_effect = analysis_data["Hub_ID"].str[1:].astype(int) % 4 * 0.25

analysis_data["Warehouse_Processing_Hours"] = (
    2.5
    + 1.0 * analysis_data["Post_Period"]
    + hub_effect
    + 0.7 * analysis_data["High_Volume"]
    + np.random.normal(0, 0.8, n)
).clip(0.5, 12)

# 8. Simulate actual delivery time
speed = analysis_data["Traffic_Level"].map({
    "Low": 50, "Medium": 35, "High": 22
})

service_effect = analysis_data["Service_Type"].map({
    "Express": -0.5, "Standard": 0, "Economy": 0.5
})

analysis_data["Travel_Hours"] = (
    analysis_data["Distance_KM"] / speed
    + 0.12 * analysis_data["Vehicle_Age"]
    + 0.8 * analysis_data["High_Volume"]
    + service_effect
    + np.random.normal(0, 1.2, n)
).clip(0.5)

analysis_data["Actual_Delivery_Hours"] = (
    analysis_data["Warehouse_Processing_Hours"]
    + analysis_data["Travel_Hours"]
)

# 9. Set promised delivery time
analysis_data["Promised_Time_Hours"] = (
    analysis_data["Distance_KM"] / 35
    + analysis_data["Service_Type"].map({
        "Express": 2, "Standard": 5, "Economy": 8
    })
)

analysis_data["On_Time"] = (
    analysis_data["Actual_Delivery_Hours"]
    <= analysis_data["Promised_Time_Hours"]
)

analysis_data["Delayed"] = (~analysis_data["On_Time"]).astype(int)

# 10. Create warehouse scan records
warehouse_scans = pd.DataFrame({
    "Shipment_ID": np.repeat(analysis_data["Shipment_ID"], 3),
    "Event_Type": ["ARRIVED", "PROCESSED", "DISPATCHED"] * n
})

warehouse_scans["Scan_Time"] = pd.to_datetime(
    np.repeat(analysis_data["Booking_Date"].values, 3)
) + pd.to_timedelta(
    np.tile([0, 2, 4], n), unit="h"
)

print("DATASET CREATED SUCCESSFULLY")
print("Shipments:", len(shipments))
print("Vehicles:", len(vehicles))
print("Routes:", len(routes))
print("Hubs:", len(hubs))
print("Warehouse scans:", len(warehouse_scans))
print("Integrated dataset:", analysis_data.shape)

display(analysis_data.head())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import ttest_ind, chi2_contingency, pearsonr
import statsmodels.formula.api as smf

# A. Overall performance
analysis_data["Week"] = (
    analysis_data["Booking_Date"].dt.to_period("W").astype(str)
)

weekly = analysis_data.groupby("Week").agg(
    Total_Shipments=("Shipment_ID", "count"),
    OTD=("On_Time", "mean"),
    Avg_Warehouse_Hours=("Warehouse_Processing_Hours", "mean")
).reset_index()

weekly["OTD_Percent"] = weekly["OTD"] * 100

print("Overall OTD:", round(analysis_data["On_Time"].mean() * 100, 2), "%")

# B. Visualize performance over time
plt.figure(figsize=(12, 5))
plt.plot(weekly["Week"], weekly["OTD_Percent"], marker="o")
plt.xticks(rotation=90)
plt.title("Weekly On-Time Delivery Performance")
plt.ylabel("On-Time Delivery (%)")
plt.xlabel("Week")
plt.grid()
plt.tight_layout()
plt.show()

# C. Hub performance
hub_performance = analysis_data.groupby("Hub_Name")["On_Time"].mean() * 100
hub_performance.sort_values().plot(kind="bar", figsize=(12, 5))
plt.title("On-Time Delivery by Hub")
plt.ylabel("OTD (%)")
plt.tight_layout()
plt.show()

# D. Warehouse processing comparison
before = analysis_data[analysis_data["Post_Period"] == 0]
after = analysis_data[analysis_data["Post_Period"] == 1]

print("H1: Warehouse processing time")
print(ttest_ind(
    before["Warehouse_Processing_Hours"],
    after["Warehouse_Processing_Hours"],
    equal_var=False
))

# E. Hub differences
hub_table = pd.crosstab(
    analysis_data["Hub_ID"], analysis_data["Delayed"]
)
print("H2: Hub performance differences")
print(chi2_contingency(hub_table))

# F. Vehicle age and delays
print("H3: Vehicle age and delivery time")
print(analysis_data[
    ["Vehicle_Age", "Actual_Delivery_Hours"]
].corr())

# G. Route differences
route_table = pd.crosstab(
    analysis_data["Route_ID"], analysis_data["Delayed"]
)
print("H4: Route differences")
print(chi2_contingency(route_table))

# H. Service type differences
service_table = pd.crosstab(
    analysis_data["Service_Type"], analysis_data["Delayed"]
)
print("H5: Service type differences")
print(chi2_contingency(service_table))

# I. Shipment volume and performance
print("H6: Volume and OTD relationship")
print(pearsonr(weekly["Total_Shipments"], weekly["OTD"]))

# J. Regression analysis
model = smf.ols(
    "Actual_Delivery_Hours ~ Warehouse_Processing_Hours + "
    "Vehicle_Age + Distance_KM + High_Traffic + High_Volume + "
    "Post_Period + C(Service_Type)",
    data=analysis_data
).fit()

print(model.summary())

# K. Early-warning view
baseline = weekly["OTD"].iloc[:8].mean()
weekly["OTD_Rolling_4"] = weekly["OTD"].rolling(4).mean()
weekly["Warning"] = (
    weekly["OTD_Rolling_4"] < baseline - 0.04
)

weekly["Warehouse_Rolling_4"] = (
    weekly["Avg_Warehouse_Hours"].rolling(4).mean()
)

print("EARLY WARNING REPORT")
display(weekly.tail(12))

print("Weeks flagged:", weekly["Warning"].sum())
```

```python

# FINAL DRIVER ANALYSIS

pre = analysis_data[analysis_data["Post_Period"] == 0].copy()
post = analysis_data[analysis_data["Post_Period"] == 1].copy()

pre_otd = pre["On_Time"].mean()
post_otd = post["On_Time"].mean()

pre_warehouse = pre["Warehouse_Processing_Hours"].mean()
post_warehouse = post["Warehouse_Processing_Hours"].mean()

pre_delivery = pre["Actual_Delivery_Hours"].mean()
post_delivery = post["Actual_Delivery_Hours"].mean()

pre_travel = pre["Travel_Hours"].mean()
post_travel = post["Travel_Hours"].mean()

# Counterfactual: estimate post-period OTD if warehouse time
# had stayed at its pre-period average.
warehouse_increase = post_warehouse - pre_warehouse

post_counterfactual = post.copy()
post_counterfactual["Counterfactual_Delivery"] = (
    post_counterfactual["Actual_Delivery_Hours"] - warehouse_increase
)

counterfactual_otd = (
    post_counterfactual["Counterfactual_Delivery"]
    <= post_counterfactual["Promised_Time_Hours"]
).mean()

otd_decline = pre_otd - post_otd
warehouse_contribution = counterfactual_otd - post_otd

if otd_decline > 0:
    contribution_percent = warehouse_contribution / otd_decline * 100
else:
    contribution_percent = np.nan

print("PERIOD COMPARISON")
print(f"Pre-period OTD: {pre_otd*100:.2f}%")
print(f"Post-period OTD: {post_otd*100:.2f}%")
print(f"OTD decline: {otd_decline*100:.2f} percentage points")
print(f"Warehouse time increase: {warehouse_increase:.2f} hours")
print(f"Average delivery time change: {post_delivery-pre_delivery:.2f} hours")
print(f"Travel time change: {post_travel-pre_travel:.2f} hours")
print()
print("WAREHOUSE COUNTERFACTUAL")
print(f"Post-period counterfactual OTD: {counterfactual_otd*100:.2f}%")
print(f"Estimated warehouse contribution: {warehouse_contribution*100:.2f} percentage points")
print(f"Share of OTD decline explained: {contribution_percent:.2f}%")

# Rank hubs, routes and service types
print("\nLOWEST OTD HUBS")
display(
    (analysis_data.groupby("Hub_Name")["On_Time"]
     .mean().mul(100).sort_values().round(2).to_frame("OTD_Percent"))
)

print("\nLOWEST OTD ROUTES")
display(
    (analysis_data.groupby("Route_ID")["On_Time"]
     .mean().mul(100).sort_values().head(10)
     .round(2).to_frame("OTD_Percent"))
)

print("\nSERVICE TYPE PERFORMANCE")
display(
    (analysis_data.groupby("Service_Type")["On_Time"]
     .mean().mul(100).round(2).to_frame("OTD_Percent"))
)
```

```python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 7))
ax.axis("off")

ax.text(0.5, 0.93, "ROOT CAUSE: DECLINE IN ON-TIME DELIVERY",
        ha="center", fontsize=15, weight="bold",
        bbox=dict(boxstyle="round,pad=0.6", fc="lightblue"))

drivers = [
    ("Warehouse processing", f"{contribution_percent:.2f}% estimated contribution"),
    ("Traffic and travel", "High traffic adds delivery time"),
    ("Vehicle age", "Weak positive association"),
    ("Hub and route variation", "Significant performance differences"),
    ("Service type", "Different delay rates"),
    ("Shipment volume", "No significant weekly relationship")
]

positions = [
    (0.12, 0.68), (0.42, 0.68), (0.72, 0.68),
    (0.12, 0.38), (0.42, 0.38), (0.72, 0.38)
]

for (name, detail), (x, y) in zip(drivers, positions):
    ax.annotate("", xy=(x+0.12, y+0.09), xytext=(0.5, 0.89),
                arrowprops=dict(arrowstyle="->", color="gray"))
    ax.text(x+0.12, y, name + "\n" + detail,
            ha="center", va="center", fontsize=9,
            bbox=dict(boxstyle="round,pad=0.6", fc="white",
                      ec="steelblue"))

ax.text(0.5, 0.12,
        "Simulated-data analysis | Counterfactual estimate, not causal proof",
        ha="center", fontsize=9, color="gray")

plt.savefig("driver_tree.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

report = f"""
# Logistics Root-Cause Analysis
## Management Report

### 1. Executive Summary

This project investigates a decline in on-time delivery (OTD)
for a simulated Kolkata freight logistics company.

OTD declined from {pre_otd*100:.2f}% to {post_otd*100:.2f}%,
a decrease of {otd_decline*100:.2f} percentage points.

Warehouse processing time increased by {warehouse_increase:.2f} hours.

### 2. Root-Cause Findings

The warehouse counterfactual estimates that maintaining the
earlier warehouse processing time would increase post-period
OTD to {counterfactual_otd*100:.2f}%.

The estimated contribution is {contribution_percent:.2f}%
of the simulated OTD decline.

This is a designed simulation result and not independent
evidence of real-world causality.

### 3. Supporting Evidence

- Hub delay rates differ significantly (chi-square test).
- Route delay rates differ significantly (chi-square test).
- Service-type delay rates differ significantly (chi-square test).
- Vehicle age has a weak positive correlation with delivery time.
- Weekly shipment volume has no significant correlation with OTD.
- Regression R-squared: {model.rsquared:.3f}.

### 4. Recommendations

| Owner | Recommended action | Proposed target |
|---|---|---|
| Warehouse Manager | Monitor arrival-to-dispatch time | Reduce average processing time by 20% |
| Fleet Manager | Introduce preventive maintenance | Reduce vehicle-related delays |
| Route Manager | Review lowest-performing routes | Improve OTD on five weakest routes |
| Hub Managers | Investigate hub bottlenecks | Improve OTD at three weakest hubs |

Targets are proposed management targets, not measured outcomes.

### 5. Early Warning System

A rolling four-week OTD monitor was developed.
Its alert threshold compares rolling OTD with the initial
eight-week baseline.

The system is a prototype and has not been validated
to provide four weeks of advance warning.

### 6. Limitations

- Data is simulated, not actual company data.
- Statistical association does not prove causation.
- Festive effects were not independently modelled.
- Early-warning lead time has not been validated.
- Route and service differences require further investigation.

### 7. Conclusion

The simulated analysis identifies warehouse processing
as the dominant designed driver of the OTD decline.
Other operational differences provide additional areas
for investigation. Real operational records are required
to validate these findings.
"""

with open("root_cause_management_report.md", "w") as f:
    f.write(report)

print("Management report created successfully.")
```
